```python
def purge_directory(path):
    try:
        # List all files and directories in the path
        items = dbutils.fs.ls(path)
        
        for item in items:
            if item.isDir():
                # Recursively delete subdirectories
                purge_directory(item.path)
                # Remove the empty directory
                dbutils.fs.rm(item.path, recurse=True)
                print(f"Deleted directory: {item.path}")
            else:
                # Delete individual files
                dbutils.fs.rm(item.path, recurse=False)
                print(f"Deleted file: {item.path}")
        
        # After deleting contents, remove the base directory if it exists
        if dbutils.fs.exists(path):
            dbutils.fs.rm(path, recurse=True)
            print(f"Purged directory: {path}")
        else:
            print(f"Path {path} does not exist or is already deleted.")
            
    except Exception as e:
        print(f"Error purging {path}: {e}")
```

```python
# Specify the path to purge
path_to_purge = "/mnt/stackoverflow/bronze/"
print(f"Starting purge of {path_to_purge}...")
purge_directory(path_to_purge)
print("Purge operation completed.")
```

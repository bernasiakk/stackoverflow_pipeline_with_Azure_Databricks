```python
storage_account_name = 'stackoverflowstorage'
container_name = 'gold'
delta_table = 'posts'

df_posts = spark.read\
					.format("delta")\
					.option("inferSchema", "true")\
					.load(f"abfss://{container_name}@{storage_account_name}.dfs.core.windows.net/{delta_table}")
     
display(df_posts)
```

```python
from pyspark.sql import functions as F

df_posts_grouped = df_posts.groupBy('owner_user_id').agg(F.count('*').alias('total_no_of_posts'))
display(df_posts_grouped)
```

```python
delta_table = 'users'

df_users = spark.read\
					.format("delta")\
					.option("inferSchema", "true")\
					.load(f"abfss://{container_name}@{storage_account_name}.dfs.core.windows.net/{delta_table}")

df_users = df_users.select(
    'display_name',
    'account_id'
)

display(df_users)
```

```python
df_joined = df_users.join(df_posts_grouped, df_posts_grouped['owner_user_id']==df_users['account_id'], how='left').filter(F.col('total_no_of_posts').isNull()).drop('owner_user_id')

display(df_joined)
```
